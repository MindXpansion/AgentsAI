```python

import os

from crewai_tools import SerperDevTool, BrowserbaseLoadTool, EXASearchTool
from crewai import Agent
from crewai import Task
from crewai import LLM

from langchain_community.llms import Ollama


from crewai import Crew, Process
```

```python

researcher_agent = Agent(
    role='Researcher',
    goal='Find and summarize the latest AI news',
    backstory="""You're a researcher at a large company.
    You're responsible for analyzing data and providing insights
    to the business.""",
    verbose=False,
    ## llm=llm_llama
    llm=LLM( model="ollama/llama3.2", base_url="http://localhost:11434" )
)
```

```python

research_task = Task(
    description="Conduct comprehensive research on the latest AI developments.",
    expected_output="A summary report highlighting key advancements in AI.",
    agent=researcher_agent,
    ## tools=[search_tool],
    async_execution=True
)
```

```python

my_crew = Crew(
    agents=[researcher_agent],
    tasks=[research_task],
    process=Process.sequential,
    verbose=True
)
```

```python

result = my_crew.kickoff()

print(result)
```


## Example 2


```python

topic = "children at school that have a cold"
```

```python

researcher_agent2 = Agent(
    role='Researcher',
    goal=f"Find and summarize comprehensive information on {topic} from reliable sources",
    backstory="""You're a researcher at a large company.
    You're responsible for analyzing data and providing insights
    to the business.""",
    verbose=False,
    ## llm=llm_llama
    llm=LLM( model="ollama/mistral:7b", base_url="http://localhost:11434" )
)
```

```python

content_writer = Agent(
    role='Writing agent',
    goal=f"transform the research findings into an engaging post and provide the html website code",
    backstory="""You're a content writer and coder at a large company.
    You're responsible for analyzing data and providing insights
    to the business that is humorous and sarcastic.""",
    verbose=False,
    ## llm=llm_llama
    llm=LLM( model="ollama/llama3.2", base_url="http://localhost:11434" )
)
```

```python

research_task = Task(
    description=f"Find and summarize comprehensive information on {topic} from reliable sources",
    expected_output="A summary report highlighting key advancements.",
    agent=researcher_agent2,
    ## tools=[search_tool],
    ## async_execution=True
)
```

```python

writing_task = Task(
    description="using text provided create a nice and humorous blog post and provide HTML code for the website",
    expected_output="A summary report highlighting key ideas in bullet form with lots of examples using lots of humor and emojis. And html code",
    agent=content_writer,
    ## tools=[search_tool],
    ## async_execution=True
)
```

```python

my_crew2 = Crew(
    agents=[  researcher_agent2, content_writer ],
    tasks=[       research_task, writing_task   ],
    process=Process.sequential,
    verbose=True
)
```

```python

result2 = my_crew2.kickoff(  inputs={"topic":topic}  ) 

print(result2)
```

```python

from crewai.tools import BaseTool
```

```python

class CalculatorTool(BaseTool):
    name: str = "Calculator tool"
    description: str = (
        "Useful to perform any mathematical calculations, like sum, minus, multiplication, division, etc. The input to this tool should be a mathematical  expression, a couple examples are `200*7` or `5000/2*10."
    )

    def _run(self, operation: str) -> int:
        # Implementation goes here
        return eval(operation)
```

```python

t = CalculatorTool()

a = t.run( "2*21" )
a
```

```python

from crewai import Agent, Task, Crew

## context = tool.run('What is natural language processing?')

tool = ScrapeWebsiteTool(website_url='https://www.rcalix.com') 

## ScrapeWebsiteTool(website_url='https://en.wikipedia.org/wiki/Artificial_intelligence')  

context = tool.run()

data_analyst = Agent(
    role='Educator',
    goal=f'Based on the context provided, answer the question - What is AI assurance? Context - {context}',
    backstory='You are a data expert',
    verbose=True,
    llm=LLM( model="ollama/llama3.2", base_url="http://localhost:11434" ),
    allow_delegation=False,
    tools=[ tool ]
)
```

```python

test_task = Task(
    description="Based on the context provided, answer the question - What is AI assurance? Context - {context}'",
    tools=[tool],
    agent=data_analyst,
    expected_output='Provide summary that addresses the question',
    verbose=True
)
```

```python

crew = Crew(
    agents=[data_analyst],
    tasks=[test_task]
)

output = crew.kickoff(  inputs={"context":context}     )
```
